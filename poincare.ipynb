#### 6. Poincaré oscillator (amplitude-phase description)
In the previous models, we described oscillators using detailed biochemical or mechanistic equations. While this is useful, it can become difficult to analyze, especially when studying synchronization or entrainment. A common simplification is to describe oscillators in terms of their **amplitude** and **phase**. The Poincaré oscillator is a minimal model that captures these features. In radial (polar) coordinates, the dynamics can be written as:


$$
\frac{dr}{dt} = \lambda \, r (A - r)
$$
$$
\frac{d\phi}{dt} = \omega = \frac{2\pi}{\tau}
$$

- $r$: instantaneous amplitude of the oscillator  
- $\phi$: phase  
- $A$: limit cycle amplitude  
- $\tau$: intrinsic period ($\omega$ is the intrinsic angular frequency)
- $\lambda$: amplitude relaxation rate, i.e. the rate at which perturbations relax to the amplitude $A$


In the code, the system is implemented in Cartesian coordinates ($x,y$), with:

$$
r = \sqrt{x^2 + y^2}
$$

and includes forcing by an external **zeitgeber** with strength $F$ and period $T$. This allows us to study **entrainment**: how an oscillator synchronizes to an external signal. The plots show oscillators with different amplitudes driven by the same zeitgeber.

Use the sliders to explore the system and think about the following:
- Two oscillators with different amplitudes (weak vs strong) are driven by the same zeitgeber. What happens as you gradually increase the forcing strength $F$?
- What happens to the weak oscillator? And to the strong oscillator? Which one entrains more easily to the zeitgeber?
- Based on these observations, how would you sketch the **Arnold tongue** for each oscillator?

Using the next code block*:
- Vary the zeitgeber strength $F$: what kinds of behaviors do you observe?
- Do you always get simple entrainment, or do more complex dynamics appear?

**Note: in this implementation, the oscillator has been slightly modified so that more complex behaviors appear at moderate parameter values, but the underlying principles remain the same.*

```python
# import needed libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint
from scipy.signal import argrelextrema

from ipywidgets import interact, interactive, fixed, interact_manual
import ipywidgets as widgets
```

```python
# define ODE model
def poincare_forced(X, t, parameters):

    x, y = X

    A = parameters['A']
    eps = parameters['eps']
    lam = parameters['lam']
    tau = parameters['tau']
    F = parameters['F']
    T= parameters['T']

    r = np.sqrt(x**2 + y**2)
    omega = 2*np.pi/tau
    Omega = 2*np.pi/T

    dxdt = lam*x*(A-r) - (omega+eps*(A-r))*y + F*np.sin(Omega*t + np.pi/2)
    dydt = lam*y*(A-r) + (omega+eps*(A-r))*x

    return [dxdt, dydt]


def simulate_poincare(A, eps, lam, F, tau, T, t):

    params = {
        'A': A,
        'eps': eps,
        'lam': lam,
        'tau': tau,
        'F': F,
        'T': T
    }

    X0 = [0. , params['A']]

    sol = odeint(poincare_forced, X0, t, args=(params,))
    return sol[:,0], sol[:,1]
```

```python
def poincare_plot(F=0.0, T=30, A_weak=0.1, A_strong=1.0):

    dt = 0.05
    n_days = 30
    t = np.arange(0,n_days*24,dt)

    # parameters
    tau = 24 
    lam = 0.1
    eps = 0.

    omega = 2*np.pi/tau
    Omega = 2*np.pi/T

    # simulate oscillators
    xw, yw = simulate_poincare(A_weak, eps, lam, F, tau, T, t)
    xs, ys = simulate_poincare(A_strong, eps, lam, F, tau, T, t)

    zeit = F*np.sin(Omega*t + np.pi/2)

    # plotting
    fig = plt.figure(figsize=(14,5))

    ax1 = fig.add_subplot(121)
    ax2 = fig.add_subplot(122)

    n_days_plot = 10 
    idx = int(24*n_days_plot/dt)

    ax1.plot(t[0:idx]/24, xw[-idx:], label='weak oscillator')
    ax1.plot(t[0:idx]/24, zeit[-idx:], '--', label='zeitgeber')
    ax1.set_title("Weak oscillator (small amplitude)")
    ax1.set_xlabel("time (days)")
    ax1.set_ylabel("x")
    ax1.grid()
    ax1.set_ylim([-2,2])

    ax2.plot(t[0:idx]/24, xs[-idx:], label='strong oscillator')
    ax2.plot(t[0:idx]/24, zeit[-idx:], '--', label='zeitgeber')
    ax2.set_title("Strong oscillator (large amplitude)")
    ax2.set_xlabel("time (days)")
    ax2.grid()
    ax2.set_ylim([-2,2])

    plt.suptitle(f"Zeitgeber strength F = {F}, Zeitgeber period T = {T}h, oscillator period = {tau}h\n")

    plt.show()


widget = interactive(
    poincare_plot,
    A_weak=(0.,1.1,0.1),
    A_strong=(0.,1.1,0.1),
    F=(0,0.2,0.025),
    T=(20,35,1)
)

display(widget)
```
